```python
alphabet = "".join([chr(i) for i in range(ord("а"), ord("я")+1)])

def encoding(ot, shift, alphabet=alphabet):
    return "".join([
        alphabet[(alphabet.find(symbol) + shift) % len(alphabet)] 
        for symbol in ot
    ])

def decoding(st, shift, alphabet=alphabet):
    return "".join([
        alphabet[(alphabet.find(symbol) - shift) % len(alphabet)] 
        for symbol in st
    ])
```

```python
file_name = "logs.txt"
# st = str.lower(input())
st = "къыуоцльуцнльощьъщпльэрчвэщмжьщщмдуэзнлчъыршръыукэшщрутнрьэурхшлчрпрэырнутщы"

with open(file_name, "w", encoding="utf-8") as f:
    
    for shift in range(1, 33):
        decoding_text = f"k={shift}: {decoding(st, shift)}"
        print(decoding_text) 
        f.write(decoding_text + "\n")

    encoding_text = encoding("гогольревизор", 11)
    print(f"{encoding_text}")
    
    report = f"""
ШИФР-ТЕКСТ (ШТ): {st}
РАСШИФРОВАННЫЙ ТЕКСТ (ОТ): {decoding(st, 11)}
КЛЮЧ: 11
АВТОР И ПРОИЗВЕДЕНИЕ (ОТ): гогольревизор
ЗАШИФРОВАННЫЕ ФАМИЛИЯ И НАЗВАНИЕ (ШТ): {encoding_text}
"""
    f.write(report)
```
