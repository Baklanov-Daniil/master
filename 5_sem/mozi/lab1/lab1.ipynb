```python
alphabet = "".join([chr(i) for i in range(ord("а"), ord("я")+1)])
# alphabet = alphabet[:6] + 'ё' + alphabet[6:]

def encoding(ot, shift, alphabet=alphabet):
    return "".join([
        alphabet[(alphabet.find(symbol) + shift) % len(alphabet)] 
        if symbol in alphabet else symbol
        for symbol in ot
    ])

def decoding(st, shift, alphabet=alphabet):
    return "".join([
        alphabet[(alphabet.find(symbol) - shift) % len(alphabet)] 
        if symbol in alphabet else symbol
        for symbol in st
    ])
```

```python
file_name = "logs.txt"

print("Выберите режим:")
print("1 - Шифрование текста")
print("2 - Расшифрование")
print("3 - Готовый текст задания")
mode = input("Введите номер режима (1 - шифрование /2-расшифрование/ 3-готовый текст из варианта):")

if mode == "1":
    st = str.lower(input("Введите текст для шифрования: "))
    shift = int(input("Введите ключ: "))
    encoded = encoding(st, shift)
    
    with open(file_name, "w", encoding="utf-8") as f:
        f.write(f"КЛЮЧ: {shift}\n")
        f.write(f"ИСХОДНЫЙ ТЕКСТ: {st}\n")
        f.write(f"ЗАШИФРОВАННЫЙ ТЕКСТ: {encoded}\n")
        print(f"КЛЮЧ: {shift}")
        print(f"ИСХОДНЫЙ ТЕКСТ: {st}")
        print(f"ЗАШИФРОВАННЫЙ ТЕКСТ: {encoded}")
    print(f"Результат сохранен в {file_name}")

elif mode == "2":
    st = str.lower(input("Введите текст для расшифрования: "))
    
    print("Выберите способ расшифрования:")
    print("1 - Ввести конкретный ключ")
    print("2 - Полный перебор всех ключей")
    method = input("Введите номер (1 - конкретный ключ/ 2 - перебор): ")
    
    with open(file_name, "w", encoding="utf-8") as f:
        if method == "1":
            shift = int(input("Введите ключ для расшифрования: "))
            decoded_text = decoding(st, shift)
            
            f.write(f"КЛЮЧ: {shift}\n")
            f.write(f"ЗАШИФРОВАННЫЙ ТЕКСТ: {st}\n")
            f.write(f"РАСШИФРОВАННЫЙ ТЕКСТ: {decoded_text}\n")
            
            print(f"КЛЮЧ: {shift}")
            print(f"РАСШИФРОВАННЫЙ ТЕКСТ: {decoded_text}")
            
        elif method == "2":
            f.write("=== ПОЛНЫЙ ПЕРЕБОР КЛЮЧЕЙ ===\n")
            for shift in range(1, 34):
                decoding_text = f"k={shift}: {decoding(st, shift)}"
                print(decoding_text)
                f.write(decoding_text + "\n")
        else:
            print("Неверный способ!")
            
    print(f"Результат сохранен в {file_name}")

elif mode == "3":
    st = "къыуоцльуцнльощьъщпльэрчвэщмжьщщмдуэзнлчъыршръыукэшщрутнрьэурхшлчрпрэырнутщы"
    
    with open(file_name, "w", encoding="utf-8") as f:
        for shift in range(1, 32):
            decoding_text = f"k={shift}: {decoding(st, shift)}"
            print(decoding_text)
            f.write(decoding_text + "\n")

        key = 11
        print(f"Ключ: k = {key}")
        encoding_text = encoding("гогольревизор", key)
        print("Автор и название: гогольревизор")
        print(f"Зашифрованный автор и название: {encoding_text}")
        
        report = f"""
ШИФР-ТЕКСТ (ШТ): {st}
РАСШИФРОВАННЫЙ ТЕКСТ (ОТ): {decoding(st, 11)}
КЛЮЧ: {key}
АВТОР И ПРОИЗВЕДЕНИЕ (ОТ): гогольревизор
ЗАШИФРОВАННЫЕ ФАМИЛИЯ И НАЗВАНИЕ (ШТ): {encoding_text}
"""
        f.write(report)
    print(f"Отчет сохранен в {file_name}")

else:
    print("Неверный режим!")
```
